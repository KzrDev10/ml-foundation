```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("Salary_Data.csv")
Column1 = df[["YearsExperience"]]
target = df["Salary"]
```

```python
X_train , X_test ,y_train ,y_test = train_test_split(Column1,target,test_size=0.3,random_state=42)

scaler = StandardScaler()
X_train_sclaed =scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python

model = LinearRegression()
model.fit(X_train_sclaed,y_train)

acc =model.score(X_train_sclaed,y_train)
weight =model.coef_
bias= model.intercept_

print(f"Your weight is {weight.item():.2f} & \nyour bias is {bias:.2f} & \nyour model training accuracy is {acc*100:.2f}%")
```

```python
prediction = model.predict(X_test_scaled)
print(f"Your model testing score is {model.score(X_test_scaled,y_test) * 100:.2f}%")
```

```python
results = pd.DataFrame({
    'Actual Salary': y_test.values,
    'Predicted Salary':prediction
})
print(results.head(5))
```
