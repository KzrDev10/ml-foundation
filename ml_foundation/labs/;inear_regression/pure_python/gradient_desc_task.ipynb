```python
import numpy as np
import pandas as pd

df = pd.read_csv('Salary_Data.csv')

x_train = np.array(df['YearsExperience'])

#target the values our model tries to predict
y_train = np.array(df['Salary'])

#total nm of examples
m = len(x_train)

print(f'Dataset loaded and with {m} examples')
```

```python
w = 0.0
b = 0.0
iteration = 1000

cost_hitory = []

#learning rate
alpha = 0.03
```

```python
for iter in range(iteration):
    total_cost = 0
    for i in range(m):
        f_wb = w*x_train[i] + b

        total_cost += (f_wb - y_train[i])**2

    cost = total_cost/(2*m)
    cost_hitory.append(cost)

    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = w*x_train[i] + b
        err = f_wb - y_train[i]

        dj_dw += err*x_train[i]
        dj_db += err

    dj_dw/=m
    dj_db/=m

    w -= (alpha * dj_dw)
    b -=(alpha* dj_db)


    if iter % 100 == 0:
        print(f"Iteration {iter:4d}: Cost = {cost:12.2f} | w = {w:8.2f} | b = {b:8.2f}")

print(f"\nFinal Parameters -> w: {w:.2f}, b: {b:.2f}")
```

```python
import matplotlib.pyplot as plt


plt.scatter(x_train, y_train, color='red', marker='x', label='Actual Data')

#  Plot the regression line using my trained w and b
y_pred = w * x_train + b
plt.plot(x_train, y_pred, color='blue', linewidth=2, label=f'Model Line (w={w:.0f}, b={b:.0f})')


plt.title('Salary vs. Experience (Linear Regression from Scratch)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
