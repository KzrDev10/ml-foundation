```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("Breast_Cancer_Data.csv")

X_m = df.drop(columns=["id","diagnosis"]).dropna(axis=1, how="all")
y = df["diagnosis"]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X_m,y,test_size=0.3,random_state=67)
# random satte can be any num most code uses 42 bc its basically a meme 
# as a supercomputer said than answer to the ult question is 42
# didnt provide the question though thats why people found it ammusing but im using 67

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
#.fit_transform  is a shortcut method that calculates parameters (like the mean and variance)
# from your data and immediately applies those changes to scale or encode the dataset in a single step.

X_test_scaled = scaler.transform(X_test)

# .tranform method modifies or scales data based on internal parameters
#  that were previously learned during the fit stage
# so basically transformers .fit once + it leaks data we dont want the model to know
# its important during testing
```

```python
model = LogisticRegression()
model.fit(X_train_scaled,y_train)

print(model.classes_)
# unlike my previous pure python logistic regression model 
# scikit learn automatically classifies the categories form the target

weight = model.coef_
bias = model.intercept_
print(f"Your weight is {weight}\nYour bias is {bias}")

test_accuracy = model.score(X_train_scaled,y_train)

print(f"\nYour model is {test_accuracy * 100:.2f} accurate")
```

```python
preds = model.predict(X_test_scaled)
print(f"Test Accuracy: {accuracy_score(y_test,preds) * 100:.2f}%")
```

```python
results = pd.DataFrame({
    'Predicted': preds,
    'Actual': y_test.values  # .values converts Pandas Series to array for clean alignment
})

print(results.head(15))
```
